### 3. **Pensando en paralelo**

**3.1 Multiplicación de dos matrices** 

**Descripción del problema:**  
Dadas dos matrices $A$ y $B$ de dimensiones $m \times n$ y $n \times p$, respectivamente, el resultado es una matriz $C$ de dimensi'on $m \times p$ donde:

\begin{equation}
C_{i,j} = \sum_{k=1}^{n} A_{i,k} \cdot B_{k,j}
\end{equation}

Cada elemento $C_{i,j}$ se calcula con una multiplicaci'on y suma de productos de la fila $i$ de $A$ con la columna $j$ de $B$.

Cada elemento $C_{i,j}$ es independiente de los dem'as y puede calcularse en paralelo. Existen diferentes estrategias:



- **Paralelización por filas/columnas:** Asignar un procesador o hilo para calcular cada fila o columna de la matriz resultante.
  
- **Paralelización por bloques:** las matrices en bloques mas pequeños y calcular las multiplicaciones en paralelo (util en supercomputadoras).
  
- **Uso de GPUs:** puede encargarse de varios elementos en paralelo aprovechando operaciones en punto flotante.


**Complejidad computacional:**  

- Método secuencial: $O(mnp)$.
  
- Método paralelo: Puede reducirse dependiendo de los recursos, alcanzando $O(mnp / P)$ si se usan $P$ procesadores.


### 3.2 **Problema de n cuerpos**

#### **Descripción del Problema:**
El problema de \( n \) cuerpos consiste en calcular las fuerzas gravitacionales entre \( n \) cuerpos en un sistema y luego actualizar sus posiciones y velocidades en función de estas fuerzas. Las fuerzas se calculan utilizando la ley de gravitación universal de Newton:

La fuerza entre los cuerpos $i$ y $j$ es $F_{i,j} = G \frac{m_i \times m_j}{r_{i,j}^2}$.

Donde $F_{i,j}$ es la fuerza entre los cuerpos $i$ y $j$, $G$ es la constante gravitacional, $m_i$ y $m_j$ son las masas de los cuerpos, y $r_{i,j}$ es la distancia entre ellos.

**Paralelización:**
El problema de \( n \) cuerpos es paralelizable porque el cálculo de las fuerzas entre cada par de cuerpos es independiente. Sin embargo, la actualización de las posiciones y velocidades de los cuerpos debe realizarse de manera secuencial después de que todas las fuerzas hayan sido calculadas

**Estrategias de paralelización:**  
1. **Paralelización por fuerza directa:** Se asigna un procesador por cuerpo para calcular la suma de fuerzas con los demás.Requiere $O(n^2)$ operaciones.s.  
2. **Método Barnes-Hut:** Aproxima la interacción usando estructuras de datos jerárquicas (quadtrees/octrees), reduciendo la complejidad $ \( O(n \log n)$\).  
3. **Uso de GPUs:** Se asignan múltiples cuerpos a diferentes unidades de procesamiento en una GPU, ejecutando cálculos en paralelo.
